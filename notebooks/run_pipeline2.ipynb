```python
# Sample heavy chain nucleotide sequence (VH region)
nt_sequence = (
    "GAGGTGCAGCTGGTGGAGTCTGGGGGAGGCTTGGTCCAGCCTGGGGGGTCCCTGAGACTCTCCTGTGCAGCCTCT"
    "GGATTCAACATCAAGGACACCTACATACACTGGGTCCGCCAGGCTCCAGGGAAGGGGCTGGAGTGGGTCGCAAGG"
    "ATCTATCCTACCAATGGTTACACACGCTATGCAGACTCCGTGAAGGGCCGATTCACCATCTCCGCCGACACCTCC"
    "AAGAACACCGCCTATCTGCAAATGAACAGCCTGAGAGCCGAGGACACGGCCGTGTATTACTGTTCAAGATGGGGG"
    "GGGGACGGCTTCTATGCTATGGACTACTGGGGTCAAGGAACCCTGGTCACCGTCTCCTCA"
)
```

```python
import json
import netam

#loading pretrained ThriftySHM
shm_model = netam.load_model("ThriftyHumV0.2-20")

#predict results
shm_results = shm_model.predict(nt_sequence)
```

```python
import os
import json

os.makedirs("/content/drive/MyDrive/antbdy_results", exist_ok=True)
output_path = "/content/drive/MyDrive/antbdy_results/shm_output.json"

with open(output_path, "w") as f:
  json.dump(shm_results, f, indent=2)

print(f"SHM output saved to {output_path}")
```
