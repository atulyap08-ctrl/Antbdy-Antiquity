```python
# Sample heavy chain nucleotide sequence (VH region)
nt_sequence = (
    "GAGGTGCAGCTGGTGGAGTCTGGGGGAGGCTTGGTCCAGCCTGGGGGGTCCCTGAGACTCTCCTGTGCAGCCTCT"
    "GGATTCAACATCAAGGACACCTACATACACTGGGTCCGCCAGGCTCCAGGGAAGGGGCTGGAGTGGGTCGCAAGG"
    "ATCTATCCTACCAATGGTTACACACGCTATGCAGACTCCGTGAAGGGCCGATTCACCATCTCCGCCGACACCTCC"
    "AAGAACACCGCCTATCTGCAAATGAACAGCCTGAGAGCCGAGGACACGGCCGTGTATTACTGTTCAAGATGGGGG"
    "GGGGACGGCTTCTATGCTATGGACTACTGGGGTCAAGGAACCCTGGTCACCGTCTCCTCA"
)
```

```python
import json
import os
from netam import pretrained
import torch

# Load pretrained ThriftySHM
shm_model = pretrained.load("ThriftyHumV0.2-20")

# Inspect the public methods available on the model
#print("Methods on shm_model:")
#print([m for m in dir(shm_model) if not m.startswith('_')])

# Sometimes netam wraps the core PyTorch module inside a .model attribute
#if hasattr(shm_model, 'model'):
#    print("\nMethods on shm_model.model:")
#    print([m for m in dir(shm_model.model) if not m.startswith('_')])

raw_predictions = shm_model.model.predictions_of_sequences([nt_sequence], encoder=shm_model.encoder)

# 3. Recursive helper to convert PyTorch Tensors to JSON-serializable types
def make_serializable(obj):
  if isinstance(obj, torch.Tensor):
    return obj.detach().cpu().tolist()
  elif isinstance(obj, dict):
    return {k: make_serializable(v) for k, v in obj.items()}
  elif isinstance(obj, (list, tuple)):
    return [make_serializable(v) for v in obj]
  return obj

# Clean the predictions
shm_results = make_serializable(raw_predictions)


# Inspect the raw output structure in the console
print("Prediction output structure:")
print(shm_results)

# Save the structured JSON to your mounted Drive
os.makedirs("/content/drive/MyDrive/antbdy_results", exist_ok=True)
output_path = "/content/drive/MyDrive/antbdy_results/shm_output.json"

with open(output_path, "w") as f:
    json.dump(shm_results, f, indent=2)

print(f"\nSHM output successfully saved to {output_path}")
```
